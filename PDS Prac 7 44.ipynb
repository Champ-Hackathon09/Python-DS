a)Write a Python program to create a scatter plot of Age and Height of 5 students.


```python
import matplotlib.pyplot as plt
import numpy as np


age = [25, 20, 18, 23, 20]
height = [160, 155, 172, 163, 180]

plt.scatter(age, height, color="blue")

plt.xlabel("Age")
plt.ylabel("Height (cm)")
plt.title("Scatter Plot of Age and Height")

plt.show()
```

b)Write a Python program to create a bar chart showing marks of 5 students.


```python
import matplotlib.pyplot as plt
import numpy as np

students = ["Aman", "Riya", "Rahul", "Priya", "Neha"]
marks = [85, 72, 90, 68, 78]

plt.bar(students, marks, color="blue")

plt.xlabel("Students")
plt.ylabel("Marks")
plt.title("Marks of 5 Students")

plt.show()
```

c)Write a Python program to create a histogram showing the distribution of marks of 10 students.


```python
import matplotlib.pyplot as plt
import numpy as np

marks = [45, 56, 67, 72, 80, 85, 90, 65, 75, 55]

plt.hist(marks)

plt.xlabel("Marks")
plt.ylabel("Number of Students")
plt.title("Distribution of Marks of 10 Students")

plt.show()
```

d)Write a Python program to create a pie chart showing the number of students in different courses.

```python
import matplotlib.pyplot as plt

courses = ["Python", "Java", "C++", "HTML"]
students = [30, 20, 15, 25]

plt.pie(students, labels=courses)

plt.title("Students in Different Courses")

plt.show()

```

e)Write a Python program to create a bar chart showing sales of 5 different products.



```python
import matplotlib.pyplot as plt

products = ["A", "B", "C", "D", "E"]
sales = [20, 35, 25, 40, 30]

plt.bar(products, sales)

plt.xlabel("Products")
plt.ylabel("Sales")
plt.title("Sales of 5 Products")

plt.show()
```
