Write a Python program using BeautifulSoup to extract and print all headings from a webpage.


```python
import requests
```

```python

from bs4 import BeautifulSoup


url = "https://www.squarespace.com/blog/website-header-examples"

response = requests.get(url)


soup = BeautifulSoup(response.text, 'html.parser')


h2_headings = soup.find_all('h2')


for heading in h2_headings:
    print(heading.text.strip())
```

Write a Python program using BeautifulSoup to extract all &lt;a&gt; tags and print their
link text and URLs.

```python

from bs4 import BeautifulSoup


url = "https://www.squarespace.com/blog/website-header-examples"


response = requests.get(url)


soup = BeautifulSoup(response.text, 'html.parser')


a_tags = soup.find_all('a')


for tag in a_tags:
    link_text = tag.text.strip()
    link_url = tag.get('href')


    if link_text and link_url:
        print(f"Text: {link_text} | URL: {link_url}")
```

```python
import seaborn as sns
import matplotlib.pyplot as plt


products = ['Laptop', 'LED TV', 'Mobile', 'Refrigerator', 'Computer']
sales = [350, 220, 480, 150, 310]


sns.set_theme(style="whitegrid")


plt.figure(figsize=(8, 5))
sns.barplot(x=products, y=sales, palette="muted")


plt.xlabel('Products', fontsize=12)
plt.ylabel('Sales Count', fontsize=12)
plt.title('Sales of 5 Different Products', fontsize=14)


plt.show()
```

Create a Seaborn scatter plot showing the relationship between students' study hours and their marks.
